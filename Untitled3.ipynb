```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a blank 200x200 RGB image
img = np.zeros((200, 200, 3), dtype=np.uint8)

# 2. Draw colored blocks and stripes
img[10:60, 10:60] = [255, 0, 0]       # Red block
img[10:60, 70:120] = [0, 255, 0]      # Green block
img[10:60, 130:180] = [0, 0, 255]     # Blue block
img[70:90, :, :] = [255, 255, 0]      # Yellow stripe
img[:, 90:100, :] = [255, 255, 255]   # White stripe
img[120:180, 20:180] = [0, 200, 255]  # Cyan block

# Display original image
plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")
plt.show()

# 3. Crop the image
cropped = img[20:150, 20:150]

plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")
plt.show()

# Flip image horizontally using negative-step slicing
flipped = img[:, ::-1, :]

plt.imshow(flipped)
plt.title("Horizontally Flipped Image")
plt.axis("off")
plt.show()

# Extract individual RGB channels
red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green_channel, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue_channel, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()

# 4. Brighten the image using float32
img_float = img.astype(np.float32)

brightened = np.clip(img_float + 50, 0, 255).astype(np.uint8)
darkened = np.clip(img_float - 50, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(brightened)
plt.title("Brightened")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(darkened)
plt.title("Darkened")
plt.axis("off")

plt.tight_layout()
plt.show()

# Find bright pixels using Boolean masking
bright_mask = np.any(img > 200, axis=2)

print("Number of bright pixels:", np.count_nonzero(bright_mask))

# Display only bright pixels
bright_pixels = np.zeros_like(img)
bright_pixels[bright_mask] = img[bright_mask]

plt.imshow(bright_pixels)
plt.title("Bright Pixels Only")
plt.axis("off")
plt.show()

# 5. Compare memory usage of uint8 and float64
img_uint8 = img
img_float64 = img.astype(np.float64)

print("Image shape:", img.shape)
print("uint8 memory:", img_uint8.nbytes, "bytes")
print("float64 memory:", img_float64.nbytes, "bytes")
print("uint8 memory:", img_uint8.nbytes / 1024, "KB")
print("float64 memory:", img_float64.nbytes / 1024, "KB")

# Display image using imshow
plt.imshow(img_uint8)
plt.title("Final Image")
plt.axis("off")
plt.show()
```
