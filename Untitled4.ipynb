```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Student': ['Aarav', 'Riya', 'Rahul', 'Sneha', 'Aditya',
                'Priya', 'Karan', 'Ananya', 'Rohan', 'Meera'],

    'Python': [85, 92, 45, 78, 66, 95, 55, 88, 72, 90],

    'Mathematics': [78, 88, 40, 82, 70, 96, 52, 91, 68, 85],

    'English': [90, 85, 55, 75, 80, 92, 60, 89, 74, 95],

    'AI_ML': [88, 95, 42, 80, 72, 98, 50, 90, 76, 93]
}

df = pd.DataFrame(data)

df
```

```python
subjects = ['Python', 'Mathematics', 'English', 'AI_ML']

df['Total'] = df[subjects].sum(axis=1)

df['Percentage'] = df['Total'] / len(subjects)

df
```

```python
def calculate_grade(percentage):

    if percentage >= 90:
        return 'A+'
    elif percentage >= 80:
        return 'A'
    elif percentage >= 70:
        return 'B'
    elif percentage >= 60:
        return 'C'
    elif percentage >= 40:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['Percentage'].apply(calculate_grade)

df
```

```python
top_students = df.sort_values(
    by='Percentage',
    ascending=False
)

print(top_students[['Student', 'Percentage', 'Grade']])
```

```python
subject_average = df[subjects].mean()

print("Subject-wise Average Marks:")
print(subject_average)
```

```python
subject_average.plot(
    kind='bar',
    color='skyblue',
    figsize=(8, 5)
)

plt.title('Subject-wise Average Performance')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df['Student'], df['Percentage'])

plt.title('Student Performance Analysis')
plt.xlabel('Students')
plt.ylabel('Percentage')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[subjects].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Subject Performance Correlation')
plt.show()
```

```python
print("STUDENT PERFORMANCE ANALYTICS REPORT")

print("------------------------------------")

print("Total Students:", len(df))

print("Class Average:",
      round(df['Percentage'].mean(), 2), "%")

print("Highest Percentage:",
      df['Percentage'].max(), "%")

print("Lowest Percentage:",
      df['Percentage'].min(), "%")

print("Topper:",
      df.loc[df['Percentage'].idxmax(), 'Student'])

print("Students Failed:",
      len(df[df['Grade'] == 'F']))
```
